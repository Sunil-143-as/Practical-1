```python
from cryptography.hazmat.primitives.asymmetric import rsa, padding
from cryptography.hazmat.primitives import serialization, hashes
```

```python
# STEP 1 : Generate RSA Private Key
# ------------------------------------------------------------
private_key = rsa.generate_private_key(
 public_exponent=65537, # Standard public exponent
 key_size=2048 # RSA Key Size
)
# ------------------------------------------------------------
# STEP 2 : Generate Public Key
# ------------------------------------------------------------
public_key = private_key.public_key()
# ------------------------------------------------------------
# STEP 3 : Display Public Key
# ------------------------------------------------------------
public_pem = public_key.public_bytes(
 encoding=serialization.Encoding.PEM,
 format=serialization.PublicFormat.SubjectPublicKeyInfo
)
print("=========== PUBLIC KEY ===========")
print(public_pem.decode())
# ------------------------------------------------------------
# STEP 4 : Display Private Key
# ------------------------------------------------------------
private_pem = private_key.private_bytes(
 encoding=serialization.Encoding.PEM,
 format=serialization.PrivateFormat.TraditionalOpenSSL,
 encryption_algorithm=serialization.NoEncryption()
)
print("\n=========== PRIVATE KEY ===========")
print(private_pem.decode())
# ------------------------------------------------------------
# STEP 5 : Read Plain Text
# ------------------------------------------------------------
message = input("Enter Message : ")
print("\nOriginal Message :", message)
# ------------------------------------------------------------
# STEP 6 : Encrypt Message
# ------------------------------------------------------------
ciphertext = public_key.encrypt(
 message.encode(),
 padding.OAEP(
 mgf=padding.MGF1(algorithm=hashes.SHA256()),
 algorithm=hashes.SHA256(),
 label=None
 )
)
print("\nEncrypted Message (Cipher Text):")
print(ciphertext)
# ------------------------------------------------------------
# STEP 7 : Decrypt Message
# ------------------------------------------------------------
plaintext = private_key.decrypt(
 ciphertext,
 padding.OAEP(
 mgf=padding.MGF1(algorithm=hashes.SHA256()),
 algorithm=hashes.SHA256(),
 label=None
 )
)
print("\nDecrypted Message :")
print(plaintext.decode())
# ------------------------------------------------------------
# STEP 8 : Verification
# ------------------------------------------------------------
if message == plaintext.decode():
 print("\nRSA Encryption and Decryption Successful.")
else:
 print("\nVerification Failed.")
```
